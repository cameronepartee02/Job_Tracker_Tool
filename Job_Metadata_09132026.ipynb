```python
df_clean = df.copy()

print(df_clean['Salary Range'])
```

```python
df_clean['Salary Range'] = df_clean['Salary Range'].apply(lambda text: text.split(':')[-1] if ':' in text else text)

df_clean['Salary Range']
```

```python
df_clean['Salary Additional'] = df_clean['Salary Range'].apply(
    lambda text: re.search(r'\(.*?\)', text).group() if ('RSU' in text or 'sign-on' in text.lower()) else None
)

df_clean[['Salary Range Low', 'Salary Range High']] = df_clean['Salary Range'].str.split('-', n=1, expand=True)

df_clean['Salary Range Low'] = df_clean['Salary Range Low'].str.replace(r'[\$,]', '', regex=True).astype(int)
df_clean['Salary Range High'] = df_clean['Salary Range High'].str.extract(r'(\$[\d,]+)')
df_clean['Salary Range High'] = df_clean['Salary Range High'].str.replace(r'[\$,]', '', regex=True).astype(int)

df_clean = df_clean.drop(columns=['Salary Range'])

df_clean
```

```python
df_clean[['Seniority Level', 'Location']]
```

```python
# --- Location: only first city/state ---
df_clean['Location Clean'] = df_clean['Location'].str.split(r'[\(;]|(?:\s+or\s+)').str[0].str.strip()

# Travel Yes/No + Allocation
df_clean['Travel'] = df_clean['Location'].str.contains('travel', case=False, na=False).map({True: 'Yes', False: 'No'})
df_clean['Travel Allocation'] = df_clean['Location'].str.extract(r'~(\d+%)')[0]
df_clean.loc[df_clean['Travel'] == 'No', 'Travel Allocation'] = None

df_clean
```

```python
# Rule-based extraction
def parse_seniority(text):
    text_lower = text.lower()

    # Try to grab an explicit years number if present
    years_match = re.search(r'(\d+)\+', text)
    years_num = years_match.group(1) if years_match else None

    if 'senior' in text_lower or 'manager' in text_lower:
        return pd.Series(['Senior', f"{years_num}+" if years_num else '5+'])
    elif 'mid-level' in text_lower or re.search(r'\bii\b', text_lower):
        return pd.Series(['Mid', f"{years_num}+" if years_num else '2+'])
    elif 'entry level' in text_lower:
        return pd.Series(['Entry Level', f"{years_num}+" if years_num else '2+'])
    elif years_num:
        return pd.Series(['Mid', f"{years_num}+"])
    else:
        return pd.Series([None, None])  # unresolved, will hardcode below

df_clean[['Seniority Category', 'Years of Experience']] = df_clean['Seniority Level'].apply(parse_seniority)

# Hardcode the two ambiguous rows by exact original text match
df_clean.loc[df_clean['Seniority Level'] == 'Not specified (no years-of-experience minimum stated)', ['Seniority Category', 'Years of Experience']] = ['Senior', '5+']
df_clean.loc[df_clean['Seniority Level'] == 'Not specified (no explicit years stated)', ['Seniority Category', 'Years of Experience']] = ['Mid', '3+']

df_clean
```

```python
# --- Job Title cleanup ---
# Strip only the parenthetical (keep comma-separated parts)
df_clean['Job Title Clean'] = df_clean['Job Title'].str.replace(r'\s*\([^)]*\)', '', regex=True).str.strip()
df_clean['Job Title Detail'] = df_clean['Job Title'].str.extract(r'\(([^)]*)\)')[0]


# --- Work Arrangement: Arrangement Type ---
def get_arrangement_type(text):
    text_lower = text.lower()
    if 'hybrid' in text_lower:
        return 'Hybrid'
    elif 'on-site' in text_lower:
        return 'On-site'
    elif 'remote' in text_lower:
        return 'Remote'
    else:
        return 'Not Specified'

df_clean['Arrangement Type'] = df_clean['Work Arrangement'].apply(get_arrangement_type)

# --- Arrangement Confidence: Stated vs Implied ---
df_clean['Arrangement Confidence'] = df_clean['Work Arrangement'].apply(
    lambda text: 'Implied' if 'not specified' in text.lower() else 'Stated'
)


# --- Fold Work Arrangement travel info into existing Travel / Travel Allocation ---

# Update Travel Yes/No to also catch travel mentioned only in Work Arrangement
df_clean['Travel'] = (
    df_clean['Location'].str.contains('travel', case=False, na=False) |
    df_clean['Work Arrangement'].str.contains('travel', case=False, na=False)
).map({True: 'Yes', False: 'No'})

# Extract the more descriptive travel range from Work Arrangement (e.g. "40-50%", "up to 80%")
df_clean['Travel Allocation WA'] = df_clean['Work Arrangement'].str.extract(r'(\d+-\d+%(?:\s*\(up to \d+%\))?|up to \d+%)', flags=re.IGNORECASE)[0]

# Prefer Work Arrangement's travel detail when present, else fall back to Location's ~XX% value
df_clean['Travel Allocation'] = df_clean['Travel Allocation WA'].combine_first(df_clean['Travel Allocation'])

# Clean up helper column
df_clean = df_clean.drop(columns=['Travel Allocation WA'])

df_clean
```

```python
# --- Build skills reference table (separate from df_clean) ---

# Combine both skill columns with a "type" flag before exploding
required = df_clean[['Job Title Clean', 'Required Skills / Tools']].copy()
required.columns = ['Job Title Clean', 'Skill']
required['Skill Type'] = 'Required'

preferred = df_clean[['Job Title Clean', 'Preferred / Nice-to-Have Skills']].copy()
preferred.columns = ['Job Title Clean', 'Skill']
preferred['Skill Type'] = 'Preferred'

skills_long = pd.concat([required, preferred], ignore_index=True)

# Split each cell on ';' into a list, then explode so each skill gets its own row
skills_long['Skill'] = skills_long['Skill'].str.split(';')
skills_long = skills_long.explode('Skill')
skills_long['Skill'] = skills_long['Skill'].str.strip()

# Drop any empty rows (e.g. blank Preferred cells)
skills_long = skills_long[skills_long['Skill'].notna() & (skills_long['Skill'] != '')]


# --- Tag each skill into a bucket category ---
def categorize_skill(skill):
    skill_lower = skill.lower()

    technical_keywords = ['sql', 'python', 'r,', ' r ', 'tableau', 'power bi', 'excel', 'etl', 'sas',
                           'matlab', 'aws', 'azure', 'gcp', 'spark', 'airflow', 'snowflake', 'redshift',
                           'ml', 'machine learning', 'model', 'llm', 'genai', 'rag', 'api', 'ci/cd',
                           'git', 'nosql', 'hadoop', 'pytorch', 'tensorflow', 'scikit', 'javascript',
                           'quicksight', 'looker', 'data studio', 'datorama', 'dbt', 'kafka', 'cortex']

    business_keywords = ['stakeholder', 'business case', 'financial modeling', 'roi', 'client',
                          'executive', 'storytelling', 'presentation', 'consulting', 'strategy',
                          'market', 'investment', 'value tracking', 'relationship management',
                          'business development', 'operating model']

    soft_keywords = ['communication', 'mentor', 'collaborate', 'leadership', 'cross-team',
                      'cross-functional', 'ambiguity', 'team experience']

    if any(kw in skill_lower for kw in technical_keywords):
        return 'Technical'
    elif any(kw in skill_lower for kw in business_keywords):
        return 'Business Acumen'
    elif any(kw in skill_lower for kw in soft_keywords):
        return 'Soft Skills'
    else:
        return 'Other/Unclassified'

skills_long['Bucket'] = skills_long['Skill'].apply(categorize_skill)

skills_long
```

```python
# --- Industry / Domain split ---
df_clean['Industry Category'] = df_clean['Industry / Domain'].str.split('/').str[0].str.strip()
df_clean['Industry Detail'] = df_clean['Industry / Domain'].str.split('/', n=1).str[1].str.strip()


# --- Company Size Tier (bucketed from raw headcount) ---
def size_tier(n):
    if n < 1000:
        return 'Startup (<1K)'
    elif n < 100000:
        return 'Mid-size (1K-100K)'
    else:
        return 'Enterprise (100K+)'

df_clean['Company Size Tier'] = df_clean['Company Size (Employees)'].apply(size_tier)


# --- Company Stage (clean tier from Company Stage / Funding) ---
def company_stage(text):
    text_lower = text.lower()
    if 'unicorn' in text_lower:
        return 'Unicorn/Startup'
    elif 'public' in text_lower:
        return 'Public'
    elif 'private' in text_lower:
        return 'Private'
    elif 'big four' in text_lower or 'established' in text_lower:
        return 'Established'
    else:
        return 'Not Specified'

df_clean['Company Stage Clean'] = df_clean['Company Stage / Funding'].apply(company_stage)

df_clean
```

```python
df_clean.loc[df_clean['Seniority Level'] == 'Consultant / Not specified (no explicit years stated)', ['Seniority Category', 'Years of Experience']] = ['Mid', '3+']
```

```python
df_clean = df_clean.drop(columns=[
    'Job Title',
    'Seniority Level',
    'Location',
    'Work Arrangement',
    'Industry / Domain',
    'Company Stage / Funding'
])
```

```python
# --- Candidate Seniority Mix: explode into long format ---

def parse_seniority_mix(text):
    # Splits "65% Senior, 30% Entry, 2% Manager, 1% VP" into [(65, 'Senior'), (30, 'Entry'), ...]
    parts = text.split(',')
    result = []
    for part in parts:
        match = re.match(r'\s*(\d+)%\s*(\w+)', part.strip())
        if match:
            result.append((int(match.group(1)), match.group(2)))
    return result

seniority_mix_long = df_clean[['Job Title Clean']].copy()
seniority_mix_long['Mix'] = df_clean['Candidate Seniority Mix (LinkedIn)'].apply(parse_seniority_mix)
seniority_mix_long = seniority_mix_long.explode('Mix')
seniority_mix_long[['Percent', 'Level']] = pd.DataFrame(seniority_mix_long['Mix'].tolist(), index=seniority_mix_long.index)
seniority_mix_long = seniority_mix_long.drop(columns=['Mix'])

seniority_mix_long
```

```python
def education_tier(text):
    text_lower = text.lower()
    if 'mba' in text_lower and ('required' in text_lower or 'in progress' in text_lower):
        return 'MBA Required'
    elif 'master' in text_lower and 'preferred' not in text_lower:
        return "Master's"
    elif 'master' in text_lower or 'mba' in text_lower:
        return "Bachelor's (Advanced Preferred)"
    elif 'bachelor' in text_lower:
        return "Bachelor's"
    else:
        return 'Not Specified'

df_clean['Education Tier'] = df_clean['Education / Certifications'].apply(education_tier)

df_clean['MBA Mentioned'] = df_clean['Education / Certifications'].str.contains('mba', case=False, na=False).map({True: 'Yes', False: 'No'})
```

```python
df_clean['Has Equity'] = df_clean['Notable Benefits'].str.contains('equity|rsu', case=False, na=False, regex=True).map({True: 'Yes', False: 'No'})
df_clean['Has Sign-on Bonus'] = df_clean['Notable Benefits'].str.contains('sign-on', case=False, na=False).map({True: 'Yes', False: 'No'})
df_clean['Has Parental Leave'] = df_clean['Notable Benefits'].str.contains('parental', case=False, na=False).map({True: 'Yes', False: 'No'})
df_clean['Benefits Specified'] = (~df_clean['Notable Benefits'].str.contains('not specified', case=False, na=False)).map({True: 'Yes', False: 'No'})

df_clean
```
