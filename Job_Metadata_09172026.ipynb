```python
# Combine skills/tools_all + key_responsibilities into skills_responsibilities
df['skills_responsibilities'] = (
    df['skills/tools_all'].fillna('') + '; ' + df['key_responsibilities'].fillna('')
)

# Clean Company Size Tier to just the category (Startup, Enterprise, Mid-size)
# This strips out anything in parentheses, e.g. "Enterprise (100K+)" -> "Enterprise"
df['Company Size Tier'] = df['Company Size Tier'].str.replace(r'\s*\(.*\)', '', regex=True).str.strip()

# Append Industry Detail onto the end of skills_responsibilities
df['skills_responsibilities'] = (
    df['skills_responsibilities'] + '; ' + df['Industry Detail'].fillna('')
)

# drop the now-redundant original columns
df = df.drop(columns=['skills/tools_all', 'key_responsibilities'])
```

```python
df = df.drop(columns=['Industry Detail'])

df.rename(columns={'Company Size Tier': 'company_size'}, inplace=True)
df.rename(columns={'Industry Category': 'industry_vertical'}, inplace=True)
```

```python
# Full reorder — list every column in the order you want
df = df[['company', 'industry_vertical', 'job_title',	'skills_responsibilities', 'years_of_experience',	'experience_level',	'travel',	'travel_allocation',	'salary_low',	'salary_high',	'est_employee_count',	'company_size', 'num_of_applicants'		]]
```

```python
# --- Reset df cleanly (drop any duplicate/old job_id columns first) ---
df = df.loc[:, ~df.columns.duplicated()]  # remove duplicate columns if any exist
if 'job_id' in df.columns:
    df = df.drop(columns=['job_id'])
df = df.reset_index().rename(columns={'index': 'job_id'})

# --- Extract fragments (long table, used internally only) ---
rows = []
for _, row in df.iterrows():
    for frag in str(row['skills_responsibilities']).split(';'):
        for sub in frag.split(','):
            sub_clean = sub.strip()
            if sub_clean:
                rows.append({
                    'job_id': row['job_id'],
                    'company': row['company'],
                    'fragment': sub_clean
                })

fragments = pd.DataFrame(rows)

# --- Pivot: fragment (skill/responsibility) x company, count of mentions ---
skill_company_counts = pd.crosstab(fragments['fragment'], fragments['company'])

skill_company_counts
```
